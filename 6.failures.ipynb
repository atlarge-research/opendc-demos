# Failures
Failures cause hosts to stop periodically. In OpenDC, failures can be simulated by providing a trace. 
This trace describes when failures occur, how long they last, and how severe they are (i.e., the number of hosts affected). 

In this demo, we will investigate the effect of failures.

#### Let's start by looking at one of the failure traces.

````python
import pandas as pd

df_failure = pd.read_parquet("failure_traces/Facebook_user_reported.parquet")

df_failure
````

- *failure_interval* determines the time between failures.
- *failure_duration* determines how long a machine cannot be used.
- *failure_intensity* determines the ratio of machines affected by the failure.

## Experiment

To use failures, users have to enable them in the experiment file. [experiments/6.failures/failure_experiment.json](experiments/6.failures/failure_experiment.json) shows an experiment that uses failures. 

A failure model is provided using the `failureModels` parameter:

```json
"failureModels": [
    {
        "type": "none"
    },
    {
        "type": "trace",
        "source": {
            "type": "named",
            "name": "failure_traces/Facebook_user_reported.parquet"
        }
    }
]
```

There are multiple types of failure models available in OpenDC. However, in this demo, we are only using a trace-based model. For more information about failure models, see [Failures](https://opendc.org/learn/documentation/Input/FailureModel). In this example, a failure model is created based on the Facebook_user_reported trace.

```python
df_task_no["total_delay"] = df_task_no["scheduling_delay"] + df_task_no["failure_delay"]
df_task_facebook["total_delay"] = df_task_facebook["scheduling_delay"] + df_task_facebook["failure_delay"]
df_task_instagram["total_delay"] = df_task_instagram["scheduling_delay"] + df_task_instagram["failure_delay"]
df_task_netflix["total_delay"] = df_task_netflix["scheduling_delay"] + df_task_netflix["failure_delay"]

delays_no = df_task_no.groupby('task_id')['total_delay'].agg('max') / 60_000
delays_facebook = df_task_facebook.groupby('task_id')['total_delay'].agg('max') / 60_000
delays_instagram = df_task_instagram.groupby('task_id')['total_delay'].agg('max') / 60_000
delays_netflix = df_task_netflix.groupby('task_id')['total_delay'].agg('max') / 60_000

print(f"The average delay without failures is: {delays_no.mean():.2f} minutes")
print(f"The average delay for Facebook is: {delays_facebook.mean():.2f} minutes")
print(f"The average delay for Instagram is: {delays_instagram.mean():.2f} minutes")
print(f"The average delay for Netflix is: {delays_netflix.mean():.2f} minutes\n")

SLA = 120

SLA_violations_no = (delays_no > SLA).sum()
SLA_violations_facebook = (delays_facebook > SLA).sum()
SLA_violations_instagram = (delays_instagram > SLA).sum()
SLA_violations_netflix = (delays_netflix > SLA).sum()

print(f"The number of SLA violations without failures is: {SLA_violations_no}")
print(f"The number of SLA violations for Facebook is: {SLA_violations_facebook}")
print(f"The number of SLA violations for Instagram is: {SLA_violations_instagram}")
print(f"The number of SLA violations for Netflix is: {SLA_violations_netflix}\n")

print(f"The percentage of SLA violations without failures is: {SLA_violations_no / len(delays_no) * 100:.2f}%")
print(f"The percentage of SLA violations for Facebook is: {SLA_violations_facebook / len(delays_facebook) * 100:.2f}%")
print(f"The percentage of SLA violations for Instagram is: {SLA_violations_instagram / len(delays_instagram) * 100:.2f}%")
print(f"The percentage of SLA violations for Netflix is: {SLA_violations_netflix / len(delays_netflix) * 100:.2f}%\n")

plt.hist(delays_no, bins=100, alpha=0.5, label='No failures')
plt.hist(delays_facebook, bins=100, alpha=0.5, label='Facebook')
plt.hist(delays_instagram, bins=100, alpha=0.5, label='Instagram')
plt.hist(delays_netflix, bins=100, alpha=0.5, label='Netflix') 

plt.axvline(x=SLA, color='r', linestyle='--', label='SLA (2 hours)')
plt.annotate('SLA (2 hours)', xy=(SLA, 0), xytext=(SLA + 100, 1), color="red")
plt.title('Impact of Horizontal Scaling on Scheduling Delay')

plt.legend()
plt.xlabel('Scheduling Delay [min]')
plt.ylabel('Frequency')

plt.show()
```
