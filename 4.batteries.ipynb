# OpenDC Demo 4: Batteries

This is the second of three demos focusing on sustainability techniques. In the previous [demo](3.horizontal_scaling.ipynb), we used OpenDC to look at the impact of downscaling the datacenter on its performance and sustainability. In this demo, we will continue by looking at batteries.

Green energy is often based on environmental phenomena. This means that its availability can vary significantly over time. For example, solar energy might be abundant during a sunny day, but is not available at night. 

In this demo, we are looking at using batteries to store energy when green energy is available, and to use this energy when only high-carbon energy is available. However, while this sounds nice, batteries have two problems: 
1. Manufacturing batteries emits carbon. This is often referred to as embodied carbon.
2. Charging large batteries can result in high power spikes that datacenters might not be able to handle.

In this demo, we explore batteries for carbon reduction.

The experiments shown in this demo are simplified versions of the experiments run for the paper [OpenDC-STEAM](https://ieeexplore.ieee.org/document/11619111). 


# Topology
To use batteries in the simulation, we need to define a battery in the topology file. 
[topologies/4.batteries/surfsara_100kWh.json](topologies/4.batteries/surfsara_100kWh.json) shows an example of a topology with a 100 kWh battery. This is defined by the following snippet:

``` json
"battery": {
    "capacity": "100 kWh",
    "chargingSpeed": "100000 W",
    "initialCharge": "0 kWh",
    "policy": {
        "type": "runningMeanPlus",
        "startingThreshold": 150.0,
        "windowSize": 168
    },
    "embodiedCarbon": 10000.0,
    "expectedLifetime": 10.0
}
```

The battery object consists of three sections:
1. The characteristics of the battery, defined by `capacity`, `chargingSpeed`, and `initialCharge`.
2. The policy that decides when the battery should charge and discharge.
3. The embodied carbon information, defined by `embodiedCarbon` and `expectedLifetime`. This is used to simulate the embodied carbon of a battery.

## Battery Policy
A datacenter uses a policy to decide when to charge and when to discharge its battery. OpenDC currently only supports policies based on the carbon intensity (CI). Four different policies are available:
1. `single`: Uses a single threshold. If the CI is lower than the threshold, the battery is charged; otherwise, the battery is discharged.
2. `double`: Similar to the `single` policy, but instead of using a single threshold, it uses an upper and a lower threshold. If the CI is lower than the lower threshold, the battery is charged. If the CI is higher than the upper threshold, the battery is discharged. If the CI is between the thresholds, the battery is idle.
3. `runningMean`: Instead of a fixed threshold, it uses the running mean of the past CI as the threshold. This works especially well for long workloads.
4. `runningMeanPlus`: Similar to the normal `runningMean`, but the performance is better because of some small tweaks.



For more information about batteries and policies, see [Battery](https://opendc.org/learn/documentation/Input/Topology/Battery).
