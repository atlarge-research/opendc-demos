# OpenDC Demo 2
### Data Center Location

In the [experiment 1](1.first_experiment.ipynb), we learned how to run OpenDC experiments and how to analyze and visualize the results to learn about the behavior of a data center. In this demo, we are comparing the behaviour of multiple data centers. More specifically, we explore how the location of a data center affects its sustainability. 

To do this, we will run the same workload on the same hardware, but in four different locations. 

# Topology

To compare data centers at different locations, we need to create more topology files, one for each location. 

**Exercise 1:** 
Create three new topology files in the topology folder of this demo: [topologies/2.datacenter_location](topologies/2.datacenter_location).

The new topologies should use the same hardware as [surfsara_NL](topologies/2.datacenter_location/surfsara_NL.json) topology, but are located at a different place. See the [carbon_traces](carbon_traces) folder to see which locations are available.     

# Experiment

Experiment files can be used to run multiple simulations, using a single file. 

In this demo we want to run a seperate simulation for each of the four data centers.

**Exercise 2:**
