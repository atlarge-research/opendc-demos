# Energy usage
fig, ax = plt.subplots(figsize=(15,5))

energy_diff_at = df_power_at.energy_usage - df_power_at.energy_usage.shift(1, fill_value=0)
energy_diff_au = df_power_au.energy_usage - df_power_au.energy_usage.shift(1, fill_value=0)
energy_diff_be = df_power_be.energy_usage - df_power_be.energy_usage.shift(1, fill_value=0)
energy_diff_nl = df_power_nl.energy_usage - df_power_nl.energy_usage.shift(1, fill_value=0)

# All four lines overlap: the location does not affect the energy usage
ax.plot(df_power_at.timestamp/1000/60/60/24, energy_diff_at / 3_600_000, label="AT") # convert J to kWh
ax.plot(df_power_au.timestamp/1000/60/60/24, energy_diff_au / 3_600_000, label="AU")
ax.plot(df_power_be.timestamp/1000/60/60/24, energy_diff_be / 3_600_000, label="BE")
ax.plot(df_power_nl.timestamp/1000/60/60/24, energy_diff_nl / 3_600_000, label="NL")

plt.title("Energy Usage per Hour at different locations")
ax.legend()

plt.xlabel("Time (days)")
plt.ylabel("Energy (kWh)")

plt.show()
