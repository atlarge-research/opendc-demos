# OpenDC Demo 3: Horizontal Scaling

In [demo 1](1.first_experiment.ipynb) and [demo 2](2.datacenter_location.ipynb), we learned the basics of how to run simulations in OpenDC and how to use Python to analyze the results. 

In the following three demos, we will use OpenDC to explore three different techniques to reduce the carbon footprint of a datacenter.
All three techniques have upsides and downsides. We use OpenDC to understand when and how to use them. 
Finally, in [demo 7](7.combining_sustainability.ipynb), we will combine the different techniques.

The experiments shown here are simplified versions of the experiments run for the paper [OpenDC-STEAM](https://ieeexplore.ieee.org/document/11619111). 

In this demo, we will focus on the first sustainability technique: horizontal scaling.
Horizontal scaling refers to reducing (or increasing) the number of machines available in a datacenter to fit a workload better.

Having a large datacenter has many benefits, such as better performance, scalability, and reliability. 
However, having too many machines can also result in many idle machines.
Idle machines still consume a significant amount of energy, while contributing nothing to the system.
This means that reducing the number of machines in a datacenter could have a positive impact on its sustainability. 

However, we also have to look at the downsides of horizontal scaling. A small datacenter cannot handle large workloads and is susceptible to workload spikes. 
To ensure good performance, we thus have to make sure we do not downscale the datacenter too much.

In this demo, we will run the same workload at different datacenter sizes to better understand how datacenter size impacts performance and sustainability.
We will use this knowledge to select the best datacenter size for a specific workload.
